```python
n = [20, 50, 100]
miu = 2.1
sigma = 0.2
alfa = 0.02
M = 500

for i in n:
    isinside = 0
    t = scipy.stats.norm.ppf(1 - alfa/2)
    for _ in range(M):
        X = np.random.normal(miu, sigma, i)
        l = np.mean(X) - t * sigma / np.sqrt(i)
        p = np.mean(X) + t * sigma / np.sqrt(i)
        if l <= miu and miu <= p:
            isinside += 1
    isinside = isinside * 100 / M
    print(f'Pokrycie dla n = {i} i sigma znanego wynosi {isinside}%')

for i in n:
    isinside = 0
    t = scipy.stats.t.ppf(1 - alfa/2, i -1)
    for _ in range(M):
        X = np.random.normal(miu, sigma, i)
        s_sigma = np.std(X)
        l = np.mean(X) - t * s_sigma / np.sqrt(i)
        p = np.mean(X) + t * s_sigma / np.sqrt(i)
        if l <= miu and miu <= p:
            isinside += 1
    isinside = isinside * 100 / M
    print(f'Pokrycie dla n = {i} i sigma nieznanego wynosi {isinside}%')
```
