```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sn
```

```python
alfa = 4
beta = 5
n = 500
M = 100
N = 1000
alfas, betas = [], []
for _ in range(M):
    X = np.random.gamma(alfa, 1/beta, n)
    S_2 = np.sum((X - np.mean(X))**2) / len(X)

    beta_est = np.mean(X)/S_2
    alfa_est = np.mean(X)**2 / S_2
    alfas.append(alfa_est)
    betas.append(beta_est)
```

```python
e1_tab, e2_tab = [], []
for _ in range(N):
    e1 = np.sum(alfas) / M - alfa
    e2 = np.sum(betas) / M - beta
    e1_tab.append(e1)
    e2_tab.append(e2)
x_arg = np.linspace(0, N, N)
plt.plot(x_arg, e1_tab)
plt.show()
plt.plot(x_arg, e2_tab)
plt.show()
```
