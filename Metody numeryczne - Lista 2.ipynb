```python
def join_vectors(A, b):
    B = A
    for i in range(len(A)):
        B[i].append(b[i])
    return B
```

```python
def Gauss_elimination_matrix(M): 
    for k in range(len(M) - 1):                 #kolumny
        biggest_num = 0
        for i in range(k, len(M)):              #wiersze
            if M[i][k] != 0 and abs(M[i][k]) > abs(biggest_num):
                biggest_num = M[i][k]
                biggest_num_index = i

        x = M[k]
        M[k] = M[biggest_num_index]
        M[biggest_num_index] = x
    
        print(M)
        for i in range(k+1, len(M)):
            a = M[i][k]/M[k][k]
            for j in range(len(M[0])):
                M[i][j] -= a * M[k][j]
        print(M)
        print('-----------------------')
    return M

W = [[0, 2, 2, 1], [3, 3, 0, 3], [1, 0, 1, 2]]
print(Gauss_elimination_matrix(W))
```

```python
def Gauss_elimination_answers(M):

    M = M[::-1]
    X = []

    for i in range(len(M)):
        x = (M[i][-1] - sum(X[::-1][j]*M[i][-1-i:-1][j] for j in range(len(X)))) / (M[i][-2 - i])
        X.append(x)

    return X[::-1]
```

```python
A = [[0, 0, 2, 1, 2], [0, 1, 0, 2, -1], [1, 2, 0, -2, 0], [0, 0, 0, -1, 1], [0, 1, -1, 1, -1]]
b = [1, 1, -4, -2, -1]
```

```python
def Gauss_elimination(A, b):
    M = join_vectors(A, b)
    return Gauss_elimination_answers(Gauss_elimination_matrix(M))

print(Gauss_elimination(A, b))
```
