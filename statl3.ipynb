```python
import numpy as np
import seaborn as sn
import matplotlib.pyplot as plt
```

```python
n = 1000

X = np.random.exponential(1/2, n)
Y = np.random.exponential(1, n)
Z = []
W = []
for i in range(len(X)):
    Z.append(max(X[i], Y[i]))
    W.append(min(X[i], Y[i]))
x_arg = np.linspace(min(X), max(X), n)


F_X = 1 - np.exp(-2 * x_arg)
F_Y = 1 - np.exp(-x_arg)
F_Z = F_X * F_Y
F_W = 1 - (1 - F_X) * (1 - F_Y)
w_arg = np.linspace(min(W), max(W), n)
```

```python
sn.ecdfplot(Z)
plt.plot(x_arg, F_Z, 'r')
plt.show()
```

```python
sn.ecdfplot(W)
plt.plot(x_arg, F_W, 'r')
plt.show()
```
