```python
import numpy as np
from scipy.interpolate import lagrange
import pandas as pd
```

```python
def Simpson_integral(f,a,b,N):
    if N % 2 == 1:
        raise ValueError("N must be an even integer.")
    dx = (b-a)/N
    x = np.linspace(a,b,N+1)
    y = f(x)
    S = dx/3 * np.sum(y[0:-1:2] + 4*y[1::2] + y[2::2])
    return S
```

### Zadanie 1

```python
v = [0, 1.0, 1.8, 2.4, 3.5, 4.4, 5.1, 6.0]
P_kw = [0, 4.7, 12.2, 19.0, 31.8, 40.1, 43.8, 43.2]
P = [i*1000 for i in P_kw]
m = 2000
v1, v2 = 1, 5
P_func = lagrange(v, P)
f_x = np.poly1d([1, 0])
def f(x):
    return f_x(x) / P_func(x)

dt = m * Simpson_integral(f, v1, v2, 4)
print(dt)
```

```python
def trapeze_method(f, a, b, N):
    x_arg = np.linspace(a, b, N+1)
    return (b - a) * (f(x_arg)[0] + f(x_arg)[-1] + 2 * sum(f(x_arg)[1::len(x_arg)-1])) / (2 * N)

def h(t):
    return t / (1 + 1/t**(4/3))

print(trapeze_method(h, 0, 1, 5))
```

```python
def Gauss_Legendre(f, a, b, N):
    def t(x):
        return (b - a) * x / 2 + (b + a) / 2
    if N == 1:
        x0, x1 = -0.5773502692, 0.5773502692
        return ((b - a) / 2) * (f(t(x0)) + f(t(x1)))
    elif N == 3:
        x0, x1, x2, x3 = -0.08611363116, -0.3399810436, 0.3399810436, 0.08611363116
        return ((b - a) / 2) * (0.3478548451*f(t(x0)) + 0.6521451549*f(t(x1)) + 0.6521451549*f(t(x2)) + 0.3478548451*f(t(x3)))
    else:
        raise ValueError("No values for this N")
    
def p(x):
    return np.log(x) / (x**2 - 2*x + 2)

print(f"Dwa węzły: {Gauss_Legendre(p, 1, np.pi, 1)}")
print(f"Cztery węzły: {Gauss_Legendre(p, 1, np.pi, 3)}")
```

### Zadanie 6

```python
def f1(x):
    return x**3 - 2*x

def f2(x):
    return np.sin(x)

def f3(x):
    return np.exp(x)
```

```python
def Df1(f, x, h):
    return (f(x+h) - f(x)) / h

def Dc2(f, x, h):
    return (f(x+h) - f(x-h)) / (2*h)

def Dc4(f, x, h):
    return (-f(x+2*h) + 8*f(x+h) - 8*f(x-h) + f(x - 2*h)) / (12*h)
```

```python
print("f1'(1) = 1")
print(f"h = 0.1 | {Df1(f1, 1, 0.1)} | {Dc2(f1, 1, 0.1)} | {Dc4(f1, 1, 0.1)}")
print(f"h = 0.01 | {Df1(f1, 1, 0.01)} | {Dc2(f1, 1, 0.01)} | {Dc4(f1, 1, 0.01)}")
print(f"h = 0.001 | {Df1(f1, 1, 0.001)} | {Dc2(f1, 1, 0.001)} | {Dc4(f1, 1, 0.001)}")
print(40*"-")
print("f2'(pi/3) = 0.5")
print(f"h = 0.1 | {Df1(f2, np.pi/3, 0.1)} | {Dc2(f2, np.pi/3, 0.1)} | {Dc4(f2, np.pi/3, 0.1)}")
print(f"h = 0.01 | {Df1(f2, np.pi/3, 0.01)} | {Dc2(f2, np.pi/3, 0.01)} | {Dc4(f2, np.pi/3, 0.01)}")
print(f"h = 0.001 | {Df1(f2, np.pi/3, 0.001)} | {Dc2(f2, np.pi/3, 0.001)} | {Dc4(f2, np.pi/3, 0.001)}")
print(40*"-")
print("f3'(0) = 1")
print(f"h = 0.1 | {Df1(f3, 0, 0.1)} | {Dc2(f3, 0, 0.1)} | {Dc4(f3, 0, 0.1)}")
print(f"h = 0.01 | {Df1(f3, 0, 0.01)} | {Dc2(f3, 0, 0.01)} | {Dc4(f3, 0, 0.01)}")
print(f"h = 0.001 | {Df1(f3, 0, 0.001)} | {Dc2(f3, 0, 0.001)} | {Dc4(f3, 0, 0.001)}")
```
